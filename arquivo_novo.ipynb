```python
import pandas as pd
import seaborn as srn
import numpy as np
from matplotlib import pyplot as plt

#analyse geral dos dados
dataset= pd.read_csv("StudentsPerformance.csv", sep=',')

print(dataset.head().to_string())

tamanho= dataset.shape
print(tamanho)

print(dataset.columns)

```

```python
dataset.columns = ['genero','raca/etnia','nivel da educacao familiar','almoco','curso de preparacao','nota em matematica','nota em leitura','nota em escrita']
```

```python
#analisando agrupamento das colunas

agrupado_0=dataset["genero"].value_counts()
print(agrupado_0)
agrupado_1=dataset["raca/etnia"].value_counts()
print(agrupado_1)
agrupado_2=dataset["nivel da educacao familiar"].value_counts()
print(agrupado_2)
agrupado_3=dataset["almoco"].value_counts()
print(agrupado_3)
agrupado_4=dataset["curso de preparacao"].value_counts()
print(agrupado_4)
agrupado_5=dataset["nota em matematica"].value_counts()
print(agrupado_5)
agrupado_6=dataset["nota em leitura"].value_counts()
print(agrupado_6)
agrupado_7=dataset["nota em escrita"].value_counts()
print(agrupado_7)
```

```python
#analise de graficos

#genero
plt.pie(agrupado_0,labels= ['female','male'],autopct='%1.1f%%')
plt.legend(title='Genero',loc="lower right")
plt.axis('equal')
plt.show()

#raça/etnia
srn.barplot(agrupado_1)

#Nivel de educação
agrupado_2.plot.bar()
plt.title("Nivel da Educação Familiar")

#Almoço
plt.pie(agrupado_3,labels= ['free/reduced','standart'],autopct='%1.1f%%')
plt.legend(title='Almoco',loc="lower right")
plt.axis('equal')
plt.show()

#Curso
plt.pie(agrupado_4,labels= ['Sim','Não'],autopct='%1.1f%%')
plt.legend(title='Fez Curso?',loc="lower right")
plt.axis('equal')
plt.show()

```
